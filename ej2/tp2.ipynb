```python
# instalamos el paquete para descargar archivos wget
#!pip install wget

from wget import download
from os import path, listdir


# descargamos la descripcion en TXT
if not path.exists("netflix_titles.txt"):
  download("https://ignaciorlando.github.io/datasets/data-science/netflix_titles.txt")
else:
  print("No vamos a bajar la descripcion porque ya existe!")

# descargamos una base de datos en CSV
if not path.exists("netflix_titles.csv"):
  download("https://ignaciorlando.github.io/datasets/data-science/netflix_titles.csv")
else:
  print("No vamos a bajar el archivo porque ya existe!")

# chequeamos si los archivos están en el directorio donde estamos parados
lista_archivos = listdir("./")
print(f"El contenido de la carpeta es: {lista_archivos}")
```

```python
import pandas as pd

# abrimos el archivo usando una función específica de pandas
raw_dataset = pd.read_csv("netflix_titles.csv", encoding='utf-8')
```

```python

raw_dataset.info()
raw_dataset.head()
```

```python
raw_dataset.info()
```

```python
# Contamos la cantidad de directores según cada valor único de RAD
cant_dir = raw_dataset['director'].nunique()

# Mostramos los resultados
print(f"Hay {cant_dir} directores en el dataset")

```

```python
#Director con mas peliculas

# filtramos el DataFrame para obtener los registros de MOVIES
registros_peliculas = raw_dataset[raw_dataset["type"] == "Movie"]

cant_peliculas = registros_peliculas.shape[0]

mayor_director = registros_peliculas.value_counts("director").sort_values(ascending=False).head(1)

print(f"{mayor_director} es el director con más películas en el dataset")
```

```python
registros_serires = raw_dataset[raw_dataset["type"] == "TV Show"]

cant_registros = registros_serires.shape[0]

print(f"Hay {cant_registros} series")
```

```python
peliculas_viejas = registros_peliculas.sort_values(by="release_year",ascending=1)

peliculas_viejas.head()
```

```python
peliculas_viejas = registros_serires.sort_values(by="release_year", ascending=1)

peliculas_viejas.head()
```

```python
pais_deseado = input("Ingrese un pais que quiera conocer")

peliculas_pais = registros_peliculas[registros_peliculas["country"] == pais_deseado]

cantidad = peliculas_pais.shape[0]

print(f"{pais_deseado} tiene {cantidad} peliculas")
```
