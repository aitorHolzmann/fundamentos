```python
serie = df['product_name'].dropna()

# 1. ¿Tiene caracteres árabes? (Rango Unicode árabe: \u0600-\u06FF)
tiene_arabe = serie.str.contains(r'[\u0600-\u06FF]', regex=True)

# 2. ¿Tiene números? (\d)
tiene_numeros = serie.str.contains(r'\d', regex=True)

# 3. ¿Tiene emojis o símbolos pictográficos?
# Rango común de emojis en Unicode
tiene_emojis = serie.str.contains(r'[\U00010000-\U0010ffff]', regex=True)

# 4. Centinelas / Nulos disfrazados típicos
patron_centinelas = r'unknown|not-applicable|n/a|none|undefined|test'
tiene_centinela = serie.str.contains(patron_centinelas, case=False)

# 5. Todo lo que NO sea A-Z, a-z ni espacios (la bolsa general que pedías)
no_az_estricto = serie.str.contains(r'[^a-zA-Z\s]', regex=True)

print("--- RESUMEN DE AUDITORÍA EN PRODUCT_NAME ---")
print(f"Total no nulos:                 {len(serie)}")
print(f"Con árabe:                      {tiene_arabe.sum()}")
print(f"Con números:                    {tiene_numeros.sum()}")
print(f"Con emojis:                     {tiene_emojis.sum()}")
print(f"Con posibles centinelas:        {tiene_centinela.sum()}")
print(f"Con algo que no sea A-Z/espacio: {no_az_estricto.sum()}")
```

```python
import pandas as pd
from IPython.display import display

def auditar_campo_texto(df, columna):
    """
    Audita texto priorizando centinelas y agrupando caracteres no A-Za-z.
    """
    if columna not in df.columns:
        print(f"Error: La columna '{columna}' no existe en el DataFrame.")
        return

    serie = df[columna].dropna().astype(str)
    total_validos = len(serie)
    nulos_reales = df[columna].isna().sum()

    print(f"==================================================")
    print(f" AUDITORÍA DE TEXTO: '{columna}'")
    print(f" Total filas: {len(df)} | Nulos (NaN): {nulos_reales} | No nulos: {total_validos}")
    print(f"==================================================\n")

    # Orden explícito de revisión
    patrones = {
        "1. Centinelas / Nulos disfrazados": r'\b(unknown|not-applicable|n/a|none|undefined|test|false|null|empty|void)\b',
        "2. Solo dígitos (códigos de barra)": r'^\d+$',
        "3. Emojis / Pictogramas": r'[\U00010000-\U0010ffff]',
        "4. Caracteres fuera de [A-Za-z] y espacios (incluye árabe, acentos, símbolos)": r'[^a-zA-Z\s]'
    }

    for nombre_prueba, regex in patrones.items():
        if "Solo dígitos" in nombre_prueba:
            mascara = serie.str.match(regex)
        else:
            mascara = serie.str.contains(regex, case=False, regex=True)

        coincidencias = serie[mascara]
        total = len(coincidencias)
        porcentaje = (total / total_validos * 100) if total_validos > 0 else 0

        print(f"\n📌 {nombre_prueba} | Coincidencias: {total} ({porcentaje:.2f}%)")

        if total > 0:
            recuentos = coincidencias.value_counts()
            with pd.option_context('display.max_rows', None, 'display.max_colwidth', None):
                display(recuentos.to_frame(name='Frecuencia'))
        else:
            print("   -> Sin coincidencias.")
        print("-" * 50)
```

```python
def histograma(columna, limites, buckets=10, ax=None):
    datos = df[columna].dropna()
    fuera_rango = (~datos.between(*limites)).sum()
    crear_figura = ax is None

    if crear_figura:
        fig, ax = plt.subplots(figsize=(8, 5))

    print(f"Valores fuera de {limites} en {columna}: {fuera_rango}")
    ax.hist(datos, bins=buckets, range=limites, edgecolor="black")
    ax.set_xlim(limites)
    ax.set_title(f"Distribución de {columna}")
    ax.set_xlabel(columna)
    ax.set_ylabel("Frecuencia")

    if crear_figura:
        fig.tight_layout()
        plt.show()

    return ax
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Definición de la muestra válida (excluyendo nulos e imposibilidades físicas)
mascara_valida = (
    df["saturated_fat_100g"].notna()
    & (df["saturated_fat_100g"] >= 0)
    & (df["saturated_fat_100g"] <= 100)
    # Criterio de consistencia opcional: no superar grasa total
    & (
        df["saturated_fat_100g"] <= df["fat_100g"]
        if "fat_100g" in df.columns
        else True
    )
)

submuestra = df[mascara_valida].copy()

# 2. Dicotomización de la variable: > 5g según consigna
submuestra["alto_grasas_sat"] = submuestra["saturated_fat_100g"] > 5

n_total = len(submuestra)
k_altos = submuestra["alto_grasas_sat"].sum()
k_bajos = n_total - k_altos
p_hat = k_altos / n_total

print(f"--- Diagnóstico Muestral ---")
print(f"Muestra efectiva analizada (n): {n_total}")
print(f"Productos con ≤ 5g: {k_bajos} ({k_bajos / n_total * 100:.2f}%)")
print(f"Productos con > 5g: {k_altos} ({p_hat * 100:.2f}%)")

# 3. Visualización con Rigor Ético
categorias = ["≤ 5 g / 100g\n(Bajo / Moderado)", "> 5 g / 100g\n(Alto en sat.)"]
porcentajes = [(k_bajos / n_total) * 100, p_hat * 100]
colores = [
    "#B0BEC5",
    "#D32F2F",
]  # Gris atenuado vs Rojo destacado (Color Focus)

fig, ax = plt.subplots(figsize=(7, 6))

barras = ax.bar(categorias, porcentajes, color=colores, width=0.55, zorder=3)

# Línea testigo del 25% (Hipótesis de referencia)
ax.axhline(
    y=25,
    color="#1976D2",
    linestyle="--",
    linewidth=1.8,
    label="Umbral de referencia H0 (25%)",
    zorder=4,
)

# Etiquetas numéricas sobre las barras
for bar in barras:
    altura = bar.get_height()
    ax.annotate(
        f"{altura:.1f}%\n({int(round(altura * n_total / 100))} u.)",
        xy=(bar.get_x() + bar.get_width() / 2, altura),
        xytext=(0, 4),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontweight="bold",
        fontsize=10,
    )

# Configuración de ejes honesta y descriptiva
ax.set_ylim(0, 100)
ax.set_ylabel("Porcentaje de productos (%)", fontsize=11)
ax.set_title(
    "Distribución de Productos según Contenido de Grasas Saturadas\n(Evaluación muestral frente al umbral del 25%)",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
ax.legend(loc="upper right", frameon=True)
ax.grid(axis="y", linestyle=":", alpha=0.6, zorder=0)

plt.tight_layout()
plt.show()
```
